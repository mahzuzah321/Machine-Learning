**Recreate the plot below. Use tips dataset. Hint: pay attention to x and y labels.** 

```python
# Define custom colors: blue and green
custom_palette = {"Male": "#4C72B0", "Female": "#55A868"}
sns.boxplot(x='day',y='tip',data=tips,hue='sex',palette=custom_palette)
```

**Recreate the plot below. Use tips dataset. Hint: pay attention to x and y labels.** 

```python
sns.jointplot(data=tips, x="total_bill", y="tip", kind="hex")
```

**Recreate the distplot given below. Use titanic dataset. Hint: pay attention to x and y labels.** 

```python
sns.distplot(titanic['fare'],bins=20,kde=False,color='green')
```

**Recreate the boxplot below using titanic dataset. Hint: pay attention to x and y labels.** 

```python
sns.boxplot(x='class',y='age',data=titanic,hue='alive',palette='coolwarm')
```

**Recreate the plot below using titanic dataset. Hint: pay attention to x and y labels.** 

```python
sns.jointplot(data=titanic, x="fare", y="age", )
```

**Recreate the plot below. Use titanic dataset. ** 

```python

custom_palette = ['#1f77b4', "lightgreen", "orange"]
sns.boxplot(x='class',y='age',data=titanic,palette=custom_palette)
```

**Can you plot how many male and female were on titanic? Separate them based on their class.**

```python
custom_colors = ["#1f77b4", "#2ca02c", "#d62728"]
sns.countplot(titanic, x="sex", hue="class",palette=custom_colors)
```

**Recreate the plot below. Use titanic dataset. Hint: Matrix plots** 

```python

import matplotlib.pyplot as plt
import pandas as pd



# Calculate the correlation matrix
correlation_matrix = titanic.corr()


sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('titanic corr()')
plt.show()
```
